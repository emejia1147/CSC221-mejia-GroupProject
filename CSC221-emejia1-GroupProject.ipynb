## Project Description

This project is inspired by NASA's Lucy mission
and its exploration of Jupiter's Trojan asteroids.

The purpose of this program is to collect orbital
and physical properties of Jupiter Trojan asteroids
using NASA/JPL's Small-Body Database REST API.

The collected data will be organized into a Pandas
DataFrame and exported to a CSV file for further
scientific analysis.

### References

- [NASA Lucy Mission](https://science.nasa.gov/mission/lucy/)

## 1. Import Python Libraries

The Requests library is used to retrieve data
from NASA's REST API, while Pandas is used to
organize and display the collected data.

```python
import pandas as pd
import requests
```

## 2. Sending the REST API Request

This section retrieves orbital properties of 100
Jupiter Trojan asteroids using NASA/JPL's Small-Body
Database REST API.

The TJN orbital classification is used to filter
the requested data. No API key is required.

```python
# NASA/JPL API endpoint
url = "https://ssd-api.jpl.nasa.gov/sbdb_query.api"

# Define the requested data
params = {
    "fields": "full_name,a,e,i,H,q,ad,per",
    "sb-class": "TJN",
    "limit": 100
}

# Send the request
response = requests.get(url, params=params, timeout=30)

# Check for errors
response.raise_for_status()

# Convert JSON response into a Python dictionary
data = response.json()

# Verify the response
print("HTTP Status:", response.status_code)
print("JSON Keys:", data.keys())
print("Records received:", len(data["data"]))
```

## 3. Creating the DataFrame

The JSON response contains asteroid records and
their corresponding field names.

Pandas is used to convert these records into a
structured table. Orbital properties are converted
into numerical data types for future analysis.

```python
# Create the DataFrame
df = pd.DataFrame(
    data["data"],
    columns=data["fields"]
)

# Convert orbital properties into numerical values
for column in df.columns:
    if column != "full_name":
        df[column] = pd.to_numeric(
            df[column], errors="coerce"
        )

# Display the first 10 records
display(df.head(10))

# Verify the dataset
print("Total rows:", len(df))
print("Total columns:", len(df.columns))
print("Missing values:")
print(df.isna().sum())
```

## 4. Data Presentation and Highlighting

Pandas styling is used to highlight the minimum
and maximum numerical values of the first 10
Jupiter Trojan asteroid records.

Green represents minimum values, while yellow
represents maximum values.

This allows the reader to easily identify variations
in orbital properties without generating a graph.

```python
# Select numerical columns
numeric_columns = ["a", "e", "i", "H", "q", "ad", "per"]

# Highlight minimum and maximum values
display(
    df.head(10).style
    .highlight_min(subset=numeric_columns, color="green")
    .highlight_max(subset=numeric_columns, color="red ")
)
```

## 5. Exporting Data to CSV

The collected Jupiter Trojan asteroid data is
exported to a CSV file using Pandas.

This file contains all 100 asteroid records and
their orbital properties for further analysis
and visualization during Part 2.

```python
# Export the DataFrame to a CSV file
df.to_csv("CSC221-GP-data.csv", index=False)

print("CSV file successfully created!")
print("Asteroid records exported:", len(df))
```
