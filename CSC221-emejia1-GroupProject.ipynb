## Project Description

This project is inspired by NASA's Lucy mission
and its exploration of Jupiter's Trojan asteroids.

The purpose of this program is to collect orbital
and physical properties of Jupiter Trojan asteroids
using NASA/JPL's Small-Body Database REST API.

The collected data will be organized into a Pandas
DataFrame and exported to a CSV file for further
scientific analysis.
